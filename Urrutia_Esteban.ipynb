```python
# Puinto 12. Se crea una funcion trigonométrica y se define el rudo como valores aleatorios entre 0 y el valor máximno de la función inicial

x = np.linspace(-2*np.pi, 2*np.pi, 300)

ruido = np.random.normal(0, 0.2, len(x))

y_ruido = np.sin(x) + ruido

y_real = np.sin(x)

plt.figure(figsize=(8,5))

plt.scatter(x, y_ruido, alpha=0.6, color='steelblue',label='sen(x) + ruido')

plt.plot(x, y_real, color='red', linewidth=2, label='sen(x)')

plt.title("Ejercicio 12")

plt.show()
```

```python
# Punto 14. Se plantean las variables y posteriormente se aplica una se trabaje sinultaneamente el seno y el coseno

x = np.linspace(-5, 5, 100)

y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = np.cos(X) + np.sin(Y)

plt.figure(figsize=(7,5))

contorno = plt.contour(X, Y, Z, cmap="viridis")

plt.colorbar(contorno)

plt.title("Ejercicio 13 - Gráfica de Contorno")

plt.show()
```

```python
# Punto 15. Para hacwer esta gráfica es importante las variables X e Y, así como su dominio

x = np.linspace(-2*np.pi, 2*np.pi, 200)

y = np.linspace(-2*np.pi, 2*np.pi, 200)

X, Y = np.meshgrid(x, y)

Z = np.sin(X)

plt.figure(figsize=(7,5))

contorno = plt.contourf(X, Y, Z, levels=20, cmap='coolwarm')

plt.colorbar(contorno)

plt.title("Ejercicio 15 - Gráfica de Contorno lleno")

plt.show()
```

```python
# Punto 16 Se generan los  1000 datos con distribución normal

datos = np.random.normal(loc=0, scale=1, size=1000)

# Crear histograma

plt.figure(figsize=(8, 5))

plt.hist(datos, bins=30, color='skyblue', edgecolor='black')

plt.title('Histograma de 1000 datos generados con distribución normal')

plt.xlabel('Valor')

plt.ylabel('Frecuencia absoluta')

plt.show()
```

```python
# Punto 17. se crean los set de datos aleatorios y con distribución normal

datos1 = np.random.normal(loc=0, scale=1, size=1000)

datos2 = np.random.normal(loc=4, scale=1.5, size=1000)

# Se crea el Histograma con las dos distribuciones

plt.figure(figsize=(8, 5))

plt.hist([datos1, datos2],bins=30, label=['Distribución 1', 'Distribución 2'])

plt.title('Dos distribuciones normales')

plt.xlabel('Valor')

plt.ylabel('Frecuencia')

plt.legend()

plt.show()
```

```python
# Punto 18. Se crran los datos para una distribución normal

datos = np.random.normal(loc=0, scale=1, size=1000)


# Calcular media

media = np.mean(datos)

# ZSe crean las diferentes distribuciones

bins_lista = [10, 30, 50]

plt.figure(figsize=(15, 4))

for i, bins in enumerate(bins_lista, 1):

  plt.subplot(1, 3, i)

  plt.hist(datos, bins=bins, color='lightgreen', edgecolor='black')

  plt.axvline(media, color='red', linestyle='--', linewidth=2, label=f'Media = {media:.2f}')

plt.title(f'Bins = {bins}')

plt.xlabel('Valor')

plt.ylabel('Frecuencia')

plt.legend()

plt.tight_layout()

plt.show()
```

```python
#Punto 20. Incialmente se crean los datos

datos1 = np.random.normal(loc=0, scale=1, size=1000)

datos2 = np.random.normal(loc=4, scale=1.5, size=1000)

# Se crean la primer figura

plt.figure(figsize=(10, 6))

plt.hist(datos1, bins=30, density=True, alpha=0.5, color='steelblue', edgecolor='black',label='Grupo A')

# Se crea la segunda figura

plt.hist(datos2, bins=30, density=True, alpha=0.5, color='tomato', edgecolor='black', label='Grupo B')


# Se empiza a fusionar las dos figuras  y se crea la fugura final

plt.axvline(np.mean(datos1), color='navy', linestyle='--', linewidth=2, label=f'Media A = {np.mean(datos1):.2f}')

plt.axvline(np.mean(datos2), color='darkred', linestyle='--', linewidth=2, label=f'Media B = {np.mean(datos2):.2f}')

plt.title('Histogramas superpuestos normalizados')

plt.xlabel('Valor')

plt.ylabel('Densidad')

plt.legend()

plt.grid(alpha=0.3)

plt.show()
```
