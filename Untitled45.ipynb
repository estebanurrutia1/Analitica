```python
mport asyncio
import speech_recognition as sr
import pyttsx3

# Inicializar TTS
engine = pyttsx3.init()

def speak(text):
    engine.say(text)
    engine.runAndWait()

def listen():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("Escuchando...")
        audio = recognizer.listen(source)

    try:
        text = recognizer.recognize_google(audio, language="es-ES")
        print(f"📝 Usuario: {text}")
        return text
    except:
        return ""

async def stt_loop(queue):
    while True:
        text = await asyncio.to_thread(listen)
        if text:
            await queue.put(text)

async def llm_loop(input_queue, output_queue):
    while True:
        text = await input_queue.get()

        # Simulación de LLM (puedes conectar OpenAI aquí)
        response = f"Escuché: {text}"

        print(f"🤖 LLM: {response}")
        await output_queue.put(response)

async def tts_loop(queue):
    while True:
        response = await queue.get()
        await asyncio.to_thread(speak, response)

async def main():
    stt_queue = asyncio.Queue()
    tts_queue = asyncio.Queue()

    await asyncio.gather(
        stt_loop(stt_queue),
        llm_loop(stt_queue, tts_queue),
        tts_loop(tts_queue)
    )

if __name__ == "__main__":
    asyncio.run(main())
```
