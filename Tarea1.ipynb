```python
!pip install opencv-python
from google.colab.patches import cv2_imshow
import cv2
import os
```

```python
ruta_imagen = "zapatos.jpg"
```

```python
imagen = cv2.imread(ruta_imagen)
```

```python
cv2_imshow(imagen)
```

```python
# Colocar la imagen en gris

gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
cv2_imshow(gris)
```

```python
# Desenfocar la imagen
desenfocada = cv2.GaussianBlur(gris, (5, 5), 0)
cv2_imshow(desenfocada)
```

```python
bordes = cv2.Canny(desenfocada, 100, 200)
cv2_imshow(bordes)
```

```python
cv2.imwrite("imagen_gris.jpg", gris)
cv2.imwrite("imagen_desenfocada.jpg", desenfocada)
cv2.imwrite("imagen_bordes.jpg", bordes)
```
